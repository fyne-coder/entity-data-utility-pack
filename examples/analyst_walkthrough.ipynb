# Entity Data Review Walkthrough

This notebook-style example shows the Python API path for analysts who want DataFrame outputs while keeping `er-review` as the underlying review engine.

```python
import pandas as pd
from er_reviewer import api

df = pd.read_csv('../sample_data/details.csv')
api.profile(df)
```

```python
api.mapping_conflicts(df, left='entity_id', right='reference_id')
```

```python
api.cluster_summary(df, cluster='entity_id', match_prefix='normalized_')
```

For very large files, prefer the CLI streaming CSV path and write outputs to disk instead of loading the whole dataset into a notebook.
